```python
import numpy as np 
import matplotlib.pyplot as plt
```

## Uniform distribution in theta and phi

```python
# Number of samples
N = 100000

# Uniformly sample points in theta and phi independently
phi = np.random.uniform(0, 2 * np.pi, N)
theta = np.random.uniform(0,  np.pi, N)
x = np.sin(theta) * np.cos(phi)

# Plot the histogram of x values
plt.hist(x, bins=100, density=True, edgecolor='black')
plt.title('Distribution of x-coordinates from independently sampled uniform distributions in theta and phi')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.show()
```

```python
y = np.sin(theta) * np.sin(phi)

# Plot the histogram of x values
plt.hist(y, bins=100, density=True, edgecolor='black')
plt.title('Distribution of y-coordinates from independently sampled uniform distributions in theta and phi')
plt.xlabel('y')
plt.ylabel('Probability Density')
plt.show()
```

```python
z =  np.cos(theta)

# Plot the histogram of z values
plt.hist(z, bins=100, density=True, edgecolor='black')
plt.title('Distribution of z-coordinates from independently sampled uniform distributions in theta and phi')
plt.xlabel('z')
plt.ylabel('Probability Density')
plt.show()
```

## Uniform distribution in cos theta and phi -> uniform distribution on unit sphere


Uniform distribution on unit sphere -> distributed uniformly in phi and **cosine theta**

```python
# Number of samples
N = 100000

# Uniformly sample points on the surface of a unit sphere
phi = np.random.uniform(0, 2 * np.pi, N)
theta = np.arccos(np.random.uniform(-1, 1, N))
x = np.sin(theta) * np.cos(phi)
y = np.sin(theta) * np.sin(phi)
z = np.cos(theta)

# Plot the histogram of x values
plt.hist(x, bins=100, density=True, edgecolor='black')
plt.title('Distribution of x-coordinates on the Unit Sphere')
plt.xlabel('x')
plt.show()
plt.hist(y, bins=100, density=True, edgecolor='black')
plt.title('Distribution of q-coordinates on the Unit Sphere')
plt.xlabel('y')
plt.show()
plt.hist(z, bins=100, density=True, edgecolor='black')
plt.title('Distribution of z-coordinates on the Unit Sphere')
plt.xlabel('z')
plt.ylabel('Probability Density')
plt.show()
plt.hist(theta, bins=100, density=True, edgecolor='black')
plt.title('Distribution of theta-coordinate')
plt.xlabel(r'$\theta$')
plt.ylabel('Probability Density')
plt.show()
plt.hist(np.cos(theta), bins=100, density=True, edgecolor='black')
plt.title(r'Distribution of cos($\theta$)')
plt.xlabel(r'$cos(\theta)$')
plt.ylabel('Probability Density')
plt.show()
plt.hist(phi, bins=100, density=True, edgecolor='black')
plt.title('Distribution of phi-coordinate')
plt.xlabel(r'$\phi$')
plt.ylabel('Probability Density')
plt.show()
```

## Checking distribution of points on a unit sphere

```python
# Checking distribution of points on a unit sphere
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Create a sphere
u = np.linspace(0, 2 * np.pi, 20)
v = np.linspace(0, np.pi, 20)
x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))

ax.plot_surface(x, y, z, color='b', alpha=0.3)

# Equally spaced points
theta = np.linspace(0, np.pi, 10)
phi = np.linspace(0, 2 * np.pi, 20)
for i in theta:
    x_points = np.sin(i) * np.cos(phi)
    y_points = np.sin(i) * np.sin(phi)
    z_points = np.cos(i)

    ax.scatter(x_points, y_points, z_points, color='r')

plt.show()

```

Conclusion: uniform distribution in theta and phi != uniform coverage of unit sphere. dA = sin\theta d\phi d\theta therefore points more closely spaced at poles than at equator if uniform in theta and phi.

```python
from mpl_toolkits.mplot3d import Axes3D
# https://arxiv.org/abs/0912.4540
def fibonacci_sphere(N):
    indices = np.arange(0, N, dtype=float) + 0.5

    phi = np.arccos(1 - 2*indices/N)
    theta = np.pi * (1 + np.sqrt(5)) * indices

    theta = theta % (2 * np.pi)  # To ensure theta is in [0, 2pi]

    return theta, phi

# Number of points
N = 1000
theta, phi = fibonacci_sphere(N)

# Convert spherical coordinates to Cartesian coordinates
x = np.sin(phi) * np.cos(theta)
y = np.sin(phi) * np.sin(theta)
z = np.cos(phi)

# Plotting
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, s=1)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Evenly Distributed Points on a Unit Sphere')
plt.show()
```
