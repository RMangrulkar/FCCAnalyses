```python
cut_data = full_data.copy().query('(P_not_light>0.999279)&(P_not_heavy>0.9989690)&(EVT_hemisEmax_n>10)')

df_bb = cut_data[cut_data['decay']=='p8_ee_Zbb_ecm91']
df_cc = cut_data[cut_data['decay']=='p8_ee_Zcc_ecm91']
df_ss = cut_data[cut_data['decay']=='p8_ee_Zss_ecm91']
df_ud = cut_data[cut_data['decay']=='p8_ee_Zud_ecm91']

fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(9,7))

# Big bins
axs[0, 0].set_title('Zbb')
axs[0,0].set_ylabel('1-P(light)')
#axs[0, 0].set_xticklabels([])
 
# Small bins
axs[0, 1].set_title('Zcc')
#axs[0, 1].set_xticklabels([])
#axs[0, 1].set_yticklabels([])


axs[1, 0].set_title('Zss')
axs[1,0].set_xlabel('1-P(heavy)')
axs[1,0].set_ylabel('1-P(light)')


axs[1, 1].set_title('Zud')
axs[1,1].set_xlabel('1-P(heavy)')
#axs[1, 1].set_yticklabels([])


h1 = axs[0, 0].hist2d(df_bb["P_not_heavy"], df_bb["P_not_light"], bins=(80, 80), cmap=plt.cm.Blues_r,vmin=0,vmax=40000000,density=True)
h2 = axs[0, 1].hist2d(df_cc["P_not_heavy"], df_cc["P_not_light"], bins=(80, 80), cmap=plt.cm.Blues_r,vmin=0,vmax=40000000,density=True)
h3 = axs[1, 0].hist2d(df_ss["P_not_heavy"], df_ss["P_not_light"], bins=(80, 80), cmap=plt.cm.Blues_r,vmin=0,vmax=40000000,density=True)
h4 = axs[1, 1].hist2d(df_ud["P_not_heavy"], df_ud["P_not_light"], bins=(80, 80), cmap=plt.cm.Blues_r,vmin=0,vmax=800000000,density=True)

# Visualizing colorbar 
fig.colorbar(h1[3], ax=axs[0, 0])
fig.colorbar(h2[3], ax=axs[0, 1], label='Density')
fig.colorbar(h3[3], ax=axs[1, 0])
fig.colorbar(h4[3], ax=axs[1, 1], label='Density')


plt.show()
```
